```python
#practice_week10_รหัสนักศึกษา.py

orders = [
    {"id": 2001, "customer": "Anan", "city": "Korat", "product": "Keyboard", "qty": 2, "price": 450},
    {"id": 2002, "customer": "Benj", "city": "Bangkok", "product": "Mouse", "qty": 5, "price": 180},
    {"id": 2003, "customer": "Chai", "city": "Bangkok", "product": "Monitor", "qty": 1, "price": 3200},
    {"id": 2004, "customer": "Duang", "city": "Korat", "product": "Mouse", "qty": 3, "price": 180},
    {"id": 2005, "customer": "Anan", "city": "Korat", "product": "Moniter", "qty": 2, "price": 3200},
    {"id": 2006, "customer": "Ekk", "city": "Bangkok", "product": "Keyboard", "qty": 4, "price": 450}
]

#1. คำนวณและเก็บลง list ของ tuple
# ลำดับข้อมูลใน tuple: (รหัสคำสั่งซื้อ, ราคาก่อน vat, vat 7%, ราคารวม vat สุทธิ)
order_summary = []

for item in orders:
    order_id = item["id"]
    subtotal = item["qty"] * item["price"] # แก้ไขจาก item[price] เป็น item["price"]
    vat = subtotal * 0.07              # ภาษีมูลค่าเพิ่ม
    net_total = subtotal + vat         # ราคารวม vat สุทธิ

    # เก็บข้อมูลลง list ของ tuple (แก้การย่อหน้าให้ตรงกับบล็อกสำหรับคำนวณ)
    order_summary.append((order_id, subtotal, vat, net_total))

# 2. ออกรายงานสรุปยอดขาย
print("=" * 65)
print(f"{'รายงานสรุปยอดขายแต่ละรายการ (Order Summary)':^65}")
print("=" * 65)
print(f"{'รหัสคำสั่งซื้อ (ID)':<20}{'ราคาก่อน VAT':>15}{'VAT (7%)':>15}{'ราคารวม VAT สุทธิ':>15}")
print("-" * 65)

grand_total_net = 0.0

# วนลูปอ่านข้อมูลจาก tuple ด้วย unpacking
for order_id, subtotal, vat, net_total in order_summary:
    grand_total_net += net_total
    print(f"{order_id:<20}{subtotal:>15,.2f}{vat:>15,.2f}{net_total:>15,.2f}")

print("-" * 65)
print(f"{'ยอดขายรวมสุทธิทั้งหมด (Grand Total)':<35}{grand_total_net:>30,.2f} บาท")
print("=" * 65)
```
